```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ast
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import (mean_absolute_error, mean_squared_error, r2_score,
                             accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, confusion_matrix,
                             ConfusionMatrixDisplay, silhouette_score)
from sklearn.cluster import KMeans, DBSCAN
from sklearn.decomposition import PCA

import os
os.makedirs('plots', exist_ok=True)

SEED = 42
np.random.seed(SEED)
print("Libraries loaded ✅")
```

```python
df = pd.read_csv('ted_main.csv')
print(f"Shape: {df.shape}")
print(f"\nColumns: {df.columns.tolist()}")
print(f"\nMissing values:\n{df.isnull().sum()}")
print(f"\nSample data:")
df.head()
```

```python
# Extract individual rating counts
rating_names = ['Funny', 'Beautiful', 'Ingenious', 'Courageous', 'Longwinded',
                'Confusing', 'Informative', 'Fascinating', 'Unconvincing',
                'Persuasive', 'Jaw-dropping', 'OK', 'Obnoxious', 'Inspiring']

df['ratings_list'] = df['ratings'].apply(ast.literal_eval)

for name in rating_names:
    col = name.lower().replace('-', '_')
    df[col] = df['ratings_list'].apply(
        lambda x, n=name: next((i['count'] for i in x if i['name'] == n), 0)
    )

rating_cols = [n.lower().replace('-', '_') for n in rating_names]

# Aggregate features
df['total_ratings'] = df[rating_cols].sum(axis=1)
df['positive_ratio'] = (df['inspiring'] + df['fascinating'] + df['beautiful'] +
                        df['ingenious'] + df['jaw_dropping'] + df['courageous']) / (df['total_ratings'] + 1)
df['negative_ratio'] = (df['longwinded'] + df['confusing'] + df['unconvincing'] +
                        df['obnoxious']) / (df['total_ratings'] + 1)

# Log transforms
df['views_log'] = np.log1p(df['views'])
df['comments_log'] = np.log1p(df['comments'])

# Date features
df['film_date'] = pd.to_datetime(df['film_date'], unit='s')
df['published_date'] = pd.to_datetime(df['published_date'], unit='s')
df['days_to_publish'] = (df['published_date'] - df['film_date']).dt.days.clip(lower=0)

# Target for classification
df['viral'] = (df['views'] > df['views'].median()).astype(int)

print(f"New columns added: {rating_cols + ['total_ratings', 'positive_ratio', 'negative_ratio', 'views_log', 'viral']}")
print(f"\nViral class balance:\n{df['viral'].value_counts()}")
print(f"\nDataset shape now: {df.shape}")
```

```python
# Descriptive statistics
print("=== Descriptive Statistics ===")
df[['views', 'comments', 'duration', 'languages', 'total_ratings', 'positive_ratio', 'negative_ratio']].describe().round(2)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df['views'], bins=50, color='steelblue', edgecolor='black')
axes[0].set_title('Views Distribution (Raw)')
axes[0].set_xlabel('Views')
axes[1].hist(df['views_log'], bins=50, color='coral', edgecolor='black')
axes[1].set_title('Views Distribution (Log Transformed)')
axes[1].set_xlabel('Log(Views)')
plt.tight_layout()
plt.savefig('plots/views_distribution.png', dpi=150)
plt.show()

# Class balance
plt.figure(figsize=(5, 4))
df['viral'].value_counts().plot(kind='bar', color=['steelblue', 'coral'], edgecolor='black')
plt.title('Viral vs Non-Viral Class Balance')
plt.xticks([0, 1], ['Non-Viral', 'Viral'], rotation=0)
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('plots/class_balance.png', dpi=150)
plt.show()

# Correlation heatmap
corr_features = ['views_log', 'comments_log', 'duration', 'languages',
                 'positive_ratio', 'negative_ratio', 'total_ratings',
                 'funny', 'inspiring', 'informative', 'fascinating', 'jaw_dropping']
corr = df[corr_features].corr()
plt.figure(figsize=(12, 9))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.savefig('plots/correlation_heatmap.png', dpi=150)
plt.show()

print("EDA plots saved ✅")
```

```python
feature_cols = ['duration', 'languages', 'comments_log', 'num_speaker',
                'days_to_publish', 'total_ratings', 'positive_ratio',
                'negative_ratio'] + rating_cols

df_model = df[feature_cols + ['views_log', 'viral']].dropna()
print(f"Modeling dataset shape: {df_model.shape}")

X = df_model[feature_cols]
y_reg = df_model['views_log']
y_clf = df_model['viral']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Regression split
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X_scaled, y_reg, test_size=0.2, random_state=SEED)

# Classification split (stratified)
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_scaled, y_clf, test_size=0.2, random_state=SEED, stratify=y_clf)

print(f"Regression   — Train: {X_train_r.shape}, Test: {X_test_r.shape}")
print(f"Classification — Train: {X_train_c.shape}, Test: {X_test_c.shape}")
print("Preprocessing done ✅")
```

```python
# Train models
lr = LinearRegression()
lr.fit(X_train_r, y_train_r)
lr_pred = lr.predict(X_test_r)

rf_reg = RandomForestRegressor(n_estimators=100, random_state=SEED)
rf_reg.fit(X_train_r, y_train_r)
rf_reg_pred = rf_reg.predict(X_test_r)

# Metrics
def reg_metrics(name, y_true, y_pred):
    return {
        'Model': name,
        'MAE': round(mean_absolute_error(y_true, y_pred), 4),
        'RMSE': round(np.sqrt(mean_squared_error(y_true, y_pred)), 4),
        'R²': round(r2_score(y_true, y_pred), 4)
    }

reg_results = pd.DataFrame([
    reg_metrics("Linear Regression", y_test_r, lr_pred),
    reg_metrics("Random Forest Regressor", y_test_r, rf_reg_pred)
])
print("=== REGRESSION RESULTS ===")
print(reg_results.to_string(index=False))

# Predicted vs Actual plots
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for ax, pred, name in zip(axes, [lr_pred, rf_reg_pred], ['Linear Regression', 'Random Forest']):
    ax.scatter(y_test_r, pred, alpha=0.3, color='steelblue', s=10)
    ax.plot([y_test_r.min(), y_test_r.max()], [y_test_r.min(), y_test_r.max()], 'r--')
    ax.set_title(f'{name}: Predicted vs Actual')
    ax.set_xlabel('Actual log(views)')
    ax.set_ylabel('Predicted log(views)')
plt.tight_layout()
plt.savefig('plots/regression_predicted_vs_actual.png', dpi=150)
plt.show()

# Feature importance
feat_imp = pd.Series(rf_reg.feature_importances_, index=feature_cols).sort_values(ascending=False).head(10)
plt.figure(figsize=(8, 5))
feat_imp.plot(kind='bar', color='steelblue', edgecolor='black')
plt.title('Top 10 Feature Importances (Random Forest Regressor)')
plt.tight_layout()
plt.savefig('plots/regression_feature_importance.png', dpi=150)
plt.show()

print("Regression done ✅")
```

```python
# Train models
log_reg = LogisticRegression(max_iter=1000, random_state=SEED)
log_reg.fit(X_train_c, y_train_c)
log_pred = log_reg.predict(X_test_c)
log_proba = log_reg.predict_proba(X_test_c)[:, 1]

rf_clf = RandomForestClassifier(n_estimators=100, random_state=SEED)
rf_clf.fit(X_train_c, y_train_c)
rf_clf_pred = rf_clf.predict(X_test_c)
rf_clf_proba = rf_clf.predict_proba(X_test_c)[:, 1]

# Metrics
def clf_metrics(name, y_true, y_pred, y_proba):
    return {
        'Model': name,
        'Accuracy': round(accuracy_score(y_true, y_pred), 4),
        'Precision': round(precision_score(y_true, y_pred), 4),
        'Recall': round(recall_score(y_true, y_pred), 4),
        'F1': round(f1_score(y_true, y_pred), 4),
        'ROC-AUC': round(roc_auc_score(y_true, y_proba), 4)
    }

clf_results = pd.DataFrame([
    clf_metrics("Logistic Regression", y_test_c, log_pred, log_proba),
    clf_metrics("Random Forest Classifier", y_test_c, rf_clf_pred, rf_clf_proba)
])
print("=== CLASSIFICATION RESULTS ===")
print(clf_results.to_string(index=False))

# Confusion matrices
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for ax, pred, name in zip(axes, [log_pred, rf_clf_pred], ['Logistic Regression', 'Random Forest']):
    cm = confusion_matrix(y_test_c, pred)
    disp = ConfusionMatrixDisplay(cm, display_labels=['Non-Viral', 'Viral'])
    disp.plot(ax=ax, colorbar=False)
    ax.set_title(name)
plt.tight_layout()
plt.savefig('plots/confusion_matrices.png', dpi=150)
plt.show()

print("Classification done ✅")
```

```python
cluster_cols = rating_cols + ['positive_ratio', 'negative_ratio', 'total_ratings']
X_clust = StandardScaler().fit_transform(df_model[cluster_cols])

# Elbow + Silhouette
inertias, sil_scores = [], []
K_range = range(2, 10)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=SEED, n_init=10)
    km.fit(X_clust)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_clust, km.labels_))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(K_range, inertias, marker='o', color='steelblue')
axes[0].set_title('Elbow Method')
axes[0].set_xlabel('Number of Clusters (k)')
axes[0].set_ylabel('Inertia')
axes[1].plot(K_range, sil_scores, marker='o', color='coral')
axes[1].set_title('Silhouette Score vs k')
axes[1].set_xlabel('Number of Clusters (k)')
axes[1].set_ylabel('Silhouette Score')
plt.tight_layout()
plt.savefig('plots/clustering_elbow_silhouette.png', dpi=150)
plt.show()

best_k = list(K_range)[np.argmax(sil_scores)]
print(f"Best k by silhouette: {best_k}")

# KMeans with best k
kmeans = KMeans(n_clusters=best_k, random_state=SEED, n_init=10)
kmeans_labels = kmeans.fit_predict(X_clust)
km_sil = silhouette_score(X_clust, kmeans_labels)
print(f"KMeans Silhouette Score (k={best_k}): {km_sil:.4f}")

# DBSCAN
dbscan = DBSCAN(eps=1.5, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_clust)
n_clusters_db = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
n_noise = (dbscan_labels == -1).sum()
print(f"DBSCAN — clusters: {n_clusters_db}, noise points: {n_noise}")
if n_clusters_db > 1:
    mask = dbscan_labels != -1
    db_sil = silhouette_score(X_clust[mask], dbscan_labels[mask])
    print(f"DBSCAN Silhouette Score: {db_sil:.4f}")

# PCA visualization
pca = PCA(n_components=2, random_state=SEED)
X_pca = pca.fit_transform(X_clust)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sc1 = axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=kmeans_labels, cmap='tab10', alpha=0.5, s=10)
axes[0].set_title(f'KMeans Clusters (k={best_k})')
axes[0].set_xlabel('PCA Component 1')
axes[0].set_ylabel('PCA Component 2')
plt.colorbar(sc1, ax=axes[0])
sc2 = axes[1].scatter(X_pca[:, 0], X_pca[:, 1], c=dbscan_labels, cmap='tab10', alpha=0.5, s=10)
axes[1].set_title('DBSCAN Clusters')
axes[1].set_xlabel('PCA Component 1')
axes[1].set_ylabel('PCA Component 2')
plt.colorbar(sc2, ax=axes[1])
plt.tight_layout()
plt.savefig('plots/clustering_pca.png', dpi=150)
plt.show()

print("Clustering done ✅")
```

```python
df_model = df_model.copy()
df_model['kmeans_cluster'] = kmeans_labels

profile = df_model.groupby('kmeans_cluster')[
    ['views_log', 'positive_ratio', 'negative_ratio', 'total_ratings',
     'inspiring', 'informative', 'fascinating', 'funny', 'languages']
].mean().round(3)

print("=== CLUSTER PROFILES ===")
print(profile.to_string())

# Visualize cluster profiles
profile[['inspiring', 'informative', 'fascinating', 'funny', 'positive_ratio', 'negative_ratio']].plot(
    kind='bar', figsize=(10, 5), edgecolor='black')
plt.title('Cluster Profiles — Average Rating Features')
plt.xlabel('Cluster')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('plots/cluster_profiles.png', dpi=150)
plt.show()

print("Cluster profile done ✅")
```

```python
print("=" * 55)
print("FULL RESULTS SUMMARY")
print("=" * 55)

print("\n--- REGRESSION ---")
print(reg_results.to_string(index=False))

print("\n--- CLASSIFICATION ---")
print(clf_results.to_string(index=False))

print("\n--- CLUSTERING ---")
print(f"KMeans (k=2) Silhouette Score: {km_sil:.4f}")
print(f"DBSCAN: {n_clusters_db} cluster(s), {n_noise} noise points")

print("\n--- CLUSTER PROFILES ---")
print(profile[['views_log', 'total_ratings', 'positive_ratio', 'inspiring', 'languages']].to_string())

print("\n✅ All results complete. Check 'plots/' for all figures.")
```

```python
from sklearn.model_selection import GridSearchCV, cross_val_score

print("=== CROSS-VALIDATION & HYPERPARAMETER TUNING ===")
print("(This may take 1-2 minutes...)\n")

# --- Random Forest Regressor tuning ---
rf_reg_params = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5]
}
rf_reg_cv = GridSearchCV(
    RandomForestRegressor(random_state=SEED),
    rf_reg_params, cv=5, scoring='r2', n_jobs=-1, verbose=0
)
rf_reg_cv.fit(X_train_r, y_train_r)
print(f"RF Regressor best params: {rf_reg_cv.best_params_}")
print(f"RF Regressor best CV R²: {rf_reg_cv.best_score_:.4f}")

# Evaluate tuned model on test set
rf_reg_tuned_pred = rf_reg_cv.best_estimator_.predict(X_test_r)
print(f"RF Regressor tuned Test R²: {r2_score(y_test_r, rf_reg_tuned_pred):.4f}")
print(f"RF Regressor tuned Test MAE: {mean_absolute_error(y_test_r, rf_reg_tuned_pred):.4f}")
print(f"RF Regressor tuned Test RMSE: {np.sqrt(mean_squared_error(y_test_r, rf_reg_tuned_pred)):.4f}")

# --- Random Forest Classifier tuning ---
print()
rf_clf_params = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5]
}
rf_clf_cv = GridSearchCV(
    RandomForestClassifier(random_state=SEED),
    rf_clf_params, cv=5, scoring='f1', n_jobs=-1, verbose=0
)
rf_clf_cv.fit(X_train_c, y_train_c)
print(f"RF Classifier best params: {rf_clf_cv.best_params_}")
print(f"RF Classifier best CV F1: {rf_clf_cv.best_score_:.4f}")

# Evaluate tuned model on test set
rf_clf_tuned_pred = rf_clf_cv.best_estimator_.predict(X_test_c)
rf_clf_tuned_proba = rf_clf_cv.best_estimator_.predict_proba(X_test_c)[:, 1]
print(f"RF Classifier tuned Test F1:      {f1_score(y_test_c, rf_clf_tuned_pred):.4f}")
print(f"RF Classifier tuned Test ROC-AUC: {roc_auc_score(y_test_c, rf_clf_tuned_proba):.4f}")
print(f"RF Classifier tuned Test Accuracy:{accuracy_score(y_test_c, rf_clf_tuned_pred):.4f}")

# --- Linear Regression 5-fold CV ---
print()
lr_cv_scores = cross_val_score(LinearRegression(), X_train_r, y_train_r, cv=5, scoring='r2')
print(f"Linear Regression 5-fold CV R²: {lr_cv_scores.mean():.4f} ± {lr_cv_scores.std():.4f}")

# --- Logistic Regression 5-fold CV ---
log_cv_scores = cross_val_score(LogisticRegression(max_iter=1000, random_state=SEED),
                                 X_train_c, y_train_c, cv=5, scoring='f1')
print(f"Logistic Regression 5-fold CV F1: {log_cv_scores.mean():.4f} ± {log_cv_scores.std():.4f}")

print("\n✅ Cross-validation done!")
```
